## Task 6: Conditional Statements

**a) Grading based on marks**

```python
marks = 85

if marks >= 90:
    grade = "A"
elif marks >= 80:
    grade = "B"
elif marks >= 70:
    grade = "C"
elif marks >= 60:
    grade = "D"
else:
    grade = "F"

print(grade)
```

**b) Checking positive, negative, or zero**

```python
number = -5

if number > 0:
    print("Positive")
elif number < 0:
    print("Negative")
else:
    print("Zero")
```

**c) Checking a leap year using nested if statements**

```python
year = 2024

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(f"{year} is a leap year")
        else:
            print(f"{year} is not a leap year")
    else:
        print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")
```

**d) Scholarship eligibility using and / or**

```python
cgpa_value = 3.7
attendance = 85

if cgpa_value > 3.5 and attendance > 80:
    print("Eligible for scholarship")
else:
    print("Not eligible for scholarship")
```

## Task 9: Classes and Inheritance

**a) Course class**

```python
class Course:
    def __init__(self, course_name, credit_hours):
        self.course_name = course_name
        self.credit_hours = credit_hours

    def display_info(self):
        print(f"Course Name: {self.course_name}")
        print(f"Credit Hours: {self.credit_hours}")
```

**b) Creating two Course objects**

```python
course1 = Course("Artificial Neural Networks", 3)
course2 = Course("Database Systems", 4)

course1.display_info()
course2.display_info()
```

**c) LabCourse class inheriting from Course**

```python
class LabCourse(Course):
    def __init__(self, course_name, credit_hours, lab_hours):
        super().__init__(course_name, credit_hours)
        self.lab_hours = lab_hours
```

**d) Overriding display_info() in LabCourse**

```python
class LabCourse(Course):
    def __init__(self, course_name, credit_hours, lab_hours):
        super().__init__(course_name, credit_hours)
        self.lab_hours = lab_hours

    def display_info(self):
        super().display_info()
        print(f"Lab Hours: {self.lab_hours}")

lab_course = LabCourse("Artificial Neural Networks Lab", 1, 3)
lab_course.display_info()
```
