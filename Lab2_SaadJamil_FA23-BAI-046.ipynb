## Task 3: Addition & Subtraction

**Creating the Term 1 and Term 2 marks matrices**

```python
term1_marks = np.array([
    [78, 85, 90],
    [65, 70, 75],
    [88, 92, 80]
])

term2_marks = np.array([
    [82, 88, 95],
    [70, 75, 80],
    [85, 90, 85]
])

print(term1_marks)
print(term2_marks)
```

**Total combined marks**

```python
total_marks = term1_marks + term2_marks
print(total_marks)
```

**Improvement (Term 2 − Term 1)**

```python
improvement = term2_marks - term1_marks
print(improvement)
```

**Identifying the student with the highest total improvement**

```python
total_improvement_per_student = np.sum(improvement, axis=1)
best_student_index = np.argmax(total_improvement_per_student)

print(total_improvement_per_student)
print(f"Student {best_student_index + 1} had the highest total improvement")
```

## Task 4: Dot Product

**Creating price and quantity vectors**

```python
prices = np.array([50, 20, 15, 100])
quantities = np.array([2, 5, 3, 1])

print(prices)
print(quantities)
```

**Total bill using the dot product**

```python
total_bill = np.dot(prices, quantities)
print(total_bill)
```

**Total bill for two customers using a single dot product**

```python
customer_quantities = np.array([
    [2, 5, 3, 1],
    [1, 2, 4, 3]
])

total_bills = np.dot(customer_quantities, prices)
print(total_bills)

# The dot product requires the number of columns in the first array (4, one per item)
# to match the number of elements in the second array (4 prices), since each output
# value is computed as the sum of element-wise products along that shared dimension.
```

## Task 5: Reshaping

**Creating a 1D array of 12 values (hours studied over 12 days)**

```python
hours_studied = np.array([2, 3, 1, 4, 2, 3, 5, 1, 2, 4, 3, 2])
print(hours_studied)
```

```python
alternate_shape = hours_studied.reshape(2, 6)
print(alternate_shape)
```

```python
try:
    invalid_shape = hours_studied.reshape(5, 3)
except ValueError as error:
    print(error)

# ValueError: cannot reshape array of size 12 into shape (5,3)
# 5 x 3 = 15, which does not match the original 12 elements, so the reshape fails.
```

## Task 8: Random Matrices

**Random decimal matrix with a fixed seed**

```python
np.random.seed(42)
decimal_matrix = np.random.rand(3, 3)
print(decimal_matrix)
```

**Random integer matrix between 10 and 100**

```python
integer_matrix = np.random.randint(10, 101, size=(3, 3))
print(integer_matrix)
```

**Random matrix from the standard normal distribution**

```python
normal_matrix = np.random.randn(4, 4)
print(normal_matrix)
```

**Re-running the seeded random generation to compare outputs**

```python
np.random.seed(42)
run1 = np.random.rand(3, 3)

np.random.seed(42)
run2 = np.random.rand(3, 3)

print(run1)
print(run2)
print(np.array_equal(run1, run2))

# Both outputs are identical because setting the same seed resets NumPy's pseudo-random
# number generator to the same internal starting state, so it produces the exact same
# sequence of "random" numbers every time it is used with that seed.
```

## Task 9: Transpose

**Creating a 3x5 performance scores matrix**

```python
performance_scores = np.array([
    [85, 88, 90, 92, 95],
    [70, 75, 78, 80, 82],
    [60, 65, 68, 70, 72]
])
print(performance_scores)
```

**Transposing and printing the new shape**

```python
transposed_scores = performance_scores.T
print(transposed_scores)
print(transposed_scores.shape)
```

**Verifying that a double transpose returns the original matrix**

```python
double_transposed = performance_scores.T.T
print(np.array_equal(performance_scores, double_transposed))
```

**Using transpose to enable a dot product between mismatched matrices**

```python
matrix_a = np.array([
    [1, 2],
    [3, 4],
    [5, 6],
    [7, 8]
])

matrix_b = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

dot_result = np.dot(matrix_a, matrix_b.T)
print(dot_result)
```
